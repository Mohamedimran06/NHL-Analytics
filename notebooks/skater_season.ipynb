```python
import pandas as pd


df = pd.read_json("data/skater_season_stats.json")
df
```

```python

import pymysql

conn = pymysql.connect(
    host='localhost',
    user='root',
    password='root',
    database='nhl'
)

cursor = conn.cursor()
```

```python
create_table_query = """
CREATE TABLE IF NOT EXISTS skater_season (
    stat_id INT PRIMARY KEY,
    player_id INT NOT NULL,
    season VARCHAR(20) NOT NULL,
    team_id INT NOT NULL,
    games_played INT DEFAULT 0,
    goals INT DEFAULT 0,
    assists INT DEFAULT 0,
    points INT DEFAULT 0,
    plus_minus INT DEFAULT 0,
    penalty_min INT DEFAULT 0,
    shots INT DEFAULT 0,
    avg_toi VARCHAR(10)
)
"""

cursor.execute(create_table_query)

insert_query = """
INSERT IGNORE INTO skater_season (
    stat_id,
    player_id,
    season,
    team_id,
    games_played,
    goals,
    assists,
    points,
    plus_minus,
    penalty_min,
    shots,
    avg_toi
)
VALUES (%s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s)
"""

columns = [
    'stat_id',
    'player_id',
    'season',
    'team_id',
    'games_played',
    'goals',
    'assists',
    'points',
    'plus_minus',
    'penalty_min',
    'shots',
    'avg_toi',
]
records = list(
    df[columns]
    .astype(object)
    .where(pd.notna(df[columns]), None)
    .itertuples(index=False, name=None)
)

cursor.executemany(insert_query, records)
conn.commit()
print(f"Inserted {cursor.rowcount} skater-season rows into skater_season")
```
