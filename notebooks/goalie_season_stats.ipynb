```python
import pandas as pd


df = pd.read_json("data/goalie_season_stats.json")
df
```

```python
import pymysql

conn = pymysql.connect(
    host='localhost',   
    user='root',
    password='root',
    database='nhl'
)

cursor = conn.cursor()
```

```python
create_table_query = """
CREATE TABLE IF NOT EXISTS goalie_season_stats (
    stat_id INT PRIMARY KEY,
    player_id INT NOT NULL,
    season VARCHAR(20) NOT NULL,
    team_id INT NOT NULL,
    games_played INT DEFAULT 0,
    wins INT DEFAULT 0,
    losses INT DEFAULT 0,
    ot_losses INT DEFAULT 0,
    save_pct DECIMAL(7, 6),
    goals_against_avg DECIMAL(7, 3),
    shutouts INT DEFAULT 0,
    saves INT
)
"""

cursor.execute(create_table_query)

insert_query = """
INSERT IGNORE INTO goalie_season_stats (
    stat_id,
    player_id,
    season,
    team_id,
    games_played,
    wins,
    losses,
    ot_losses,
    save_pct,
    goals_against_avg,
    shutouts,
    saves
)
VALUES (%s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s)
"""

columns = [
    'stat_id',
    'player_id',
    'season',
    'team_id',
    'games_played',
    'wins',
    'losses',
    'ot_losses',
    'save_pct',
    'goals_against_avg',
    'shutouts',
    'saves',
]
records = list(
    df[columns]
    .astype(object)
    .where(pd.notna(df[columns]), None)
    .itertuples(index=False, name=None)
)

cursor.executemany(insert_query, records)
conn.commit()
print(f"Inserted {cursor.rowcount} goalie-season rows into goalie_season_stats")
```
